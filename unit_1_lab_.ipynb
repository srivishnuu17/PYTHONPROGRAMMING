```python

# Scientific Calculator for Two Numbers

num1 = float(input("Enter First Number: "))
num2 = float(input("Enter Second Number: "))

print("\n========== CALCULATOR ==========")

print(f"Addition        : {num1 + num2}")
print(f"Subtraction     : {num1 - num2}")
print(f"Multiplication  : {num1 * num2}")

if num2 != 0:
    print(f"Division        : {num1 / num2}")
    print(f"Floor Division  : {num1 // num2}")
    print(f"Modulus         : {num1 % num2}")
else:
    print("Division        : Not possible")
    print("Floor Division  : Not possible")
    print("Modulus         : Not possible")

print(f"Exponentiation  : {num1 ** num2}")
print("================================")
```

```python

# Mathematical Calculator using Built-in Functions and Math Module

import math

number = float(input("Enter a number: "))

print("\n===== MATHEMATICAL OPERATIONS =====")

print(f"Number          : {number}")
print(f"Square          : {number ** 2}")
print(f"Cube            : {number ** 3}")

if number >= 0:
    print(f"Square Root     : {math.sqrt(number)}")
else:
    print("Square Root     : Not defined for negative numbers")

print(f"Ceiling Value   : {math.ceil(number)}")
print(f"Floor Value     : {math.floor(number)}")
print(f"Absolute Value  : {abs(number)}")
print(f"Type            : {type(number)}")
print(f"Memory Address  : {id(number)}")

print("===================================")
```

```python

# Electricity Bill Generator

print("========== ELECTRICITY BILL GENERATOR ==========")

consumer_name = input("Enter Consumer Name: ")
consumer_id = input("Enter Consumer ID: ")

previous_reading = float(input("Enter Previous Meter Reading (kWh): "))
current_reading = float(input("Enter Current Meter Reading (kWh): "))
cost_per_unit = float(input("Enter Cost per Unit (Rs): "))

# Calculations
units = current_reading - previous_reading

if units < 0 or cost_per_unit < 0:
    print("Invalid meter reading or cost per unit!")
else:
    energy_charge = units * cost_per_unit
    electricity_duty = energy_charge * 0.05
    fixed_charge = 100

    net_bill = energy_charge + electricity_duty + fixed_charge

    # Display Electricity Bill
    print("\n============= ELECTRICITY BILL =============")
    print(f"Consumer Name        : {consumer_name}")
    print(f"Consumer ID          : {consumer_id}")
    print("---------------------------------------------")
    print(f"Previous Reading     : {previous_reading:.2f} kWh")
    print(f"Current Reading      : {current_reading:.2f} kWh")
    print(f"Units Consumed       : {units:.2f} kWh")
    print("---------------------------------------------")
    print(f"Energy Charge        : Rs. {energy_charge:.2f}")
    print(f"Electricity Duty 5%  : Rs. {electricity_duty:.2f}")
    print(f"Fixed Meter Charge   : Rs. {fixed_charge:.2f}")
    print("---------------------------------------------")
    print(f"Net Bill Amount      : Rs. {net_bill:.2f}")
    print("=============================================")
```
